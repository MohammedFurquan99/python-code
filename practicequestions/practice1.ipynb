```python
#sorts in ascending order 
list = [1,2,4,5,7,2,5,97,4,7,6,9,]
print(list.sort())
print(list)
```

```python
# sorts in decending order
nums = [1,2,4,5,7,2,5,97,4,7,6,9,100000]
print(nums.sort(reverse=True))
print(nums)
```
